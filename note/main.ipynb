```python
def butter_lowpass(lowcut, fs, order=4):
    '''バターワースローパスフィルタを設計する関数
    '''
    nyq = 0.5 * fs
    low = lowcut / nyq
    b, a = signal.butter(order, low, btype='low')
    return b, a
```

```python
def butter_lowpass_filter(x, lowcut, fs, order=4):
    '''データにローパスフィルタをかける関数
    '''
    b, a = butter_lowpass(lowcut, fs, order=order)
    y = signal.filtfilt(b, a, x)
    return y
```

```python
#uvマップ内積
import numpy as np
import matplotlib.pyplot as plt
import cv2
from pathlib import Path
from tqdm import tqdm
from mpl_toolkits.axes_grid1 import ImageGrid
import copy # ★背景白化処理に必要
from scipy.ndimage import median_filter # ★空間フィルタ用に追加

# --- 設定項目 ---
# UV展開後の .npy ディレクトリ (Makeup)
UV_MAKEUP_DIR = Path(r"C:\Users\hil\Documents\cosme\output_uvs_texture\fund\exp004_20260325_14_34_fund3_1_p70_uvs")

# 参照信号を取得する領域 (Region of Interest)
ROI_X = 80   # 左上X 
ROI_Y = 130  # 左上Y 
ROI_W = 15   # 幅
ROI_H = 15   # 高さ

# 解析フレーム
START_F = 0
END_F = 300
# ----------------

def check_roi_position(img_dir: Path, x, y, w, h):
    """
    ROI(参照領域)が顔のどのあたりにあるかを確認表示する関数
    """
    files = sorted(list(img_dir.glob("*.npy")))
    if not files:
        print("ファイルが見つかりません。")
        return

    # 1枚読み込み
    img = np.load(files[0])
    
    # 表示用に正規化 (0-255)
    disp_img = np.clip(img, 0, 255).astype(np.uint8).copy()
    
    # ROIを赤枠で描画
    cv2.rectangle(disp_img, (x, y), (x+w, y+h), (0, 0, 255), 2)
    
    # RGB変換して表示
    disp_img = cv2.cvtColor(disp_img, cv2.COLOR_BGR2RGB)
    
    plt.figure(figsize=(4, 4))
    plt.imshow(disp_img)
    plt.title(f"Reference Area (ROI)\nx={x}, y={y}")
    plt.axis('off')
    plt.show()

def get_signal_from_uv_roi(img_dir: Path, start_f, end_f, x, y, w, h):
    """
    UV展開画像群の特定領域(ROI)から緑色信号(AC成分)を抽出
    """
    files = sorted(list(img_dir.glob("*.npy")))
    files = files[start_f : min(len(files), end_f)]
    
    signal = []
    accum_val = None
    
    print(f"信号抽出中 (UV Map ROI): {img_dir.name}")
    
    for f_path in tqdm(files):
        # 画像読み込み (float)
        img = np.load(f_path)
        
        # ★修正箇所1: 画像全体(Gチャンネル)に3x3空間メディアンフィルタを適用
        raw_g = img[:, :, 1]
        filtered_g = median_filter(raw_g, size=(3, 3))
        
        # ROI切り出し (フィルタ後の画像から)
        roi = filtered_g[y:y+h, x:x+w]
        
        # 中央値(Median)を使用
        val = np.median(roi)
        
        # ローカットフィルタ (AC成分抽出)
        if accum_val is None:
            accum_val = val
            ac = 0.0
        else:
            accum_val = accum_val * 0.95 + val * 0.05
            ac = val - accum_val
            
        signal.append(ac)
        
    return np.array(signal)

def visualize_correlation(uv_dir: Path, ref_signal: np.ndarray, start_f: int):
    """
    UV画像群と参照信号の内積マップを作成
    """
    files = sorted(list(uv_dir.glob("*.npy")))
    # 信号長に合わせる
    files = files[start_f : start_f + len(ref_signal)]
    
    h_map, w_map = 256, 256
    accum_face = np.zeros((h_map, w_map), dtype=np.float64)
    inner_map = np.zeros((h_map, w_map), dtype=np.float64)
    
    print("内積マップ計算中...")
    
    for i, f_path in enumerate(tqdm(files)):
        img = np.load(f_path)
        
        # ★修正箇所2: マップ計算時にも同様に3x3空間メディアンフィルタを適用
        raw_g = img[:, :, 1]
        filtered_g = median_filter(raw_g, size=(3, 3))
        
        face_g = filtered_g.astype(np.float64)
        
        if i == 0:
            accum_face = face_g
            face_low_cut = np.zeros_like(face_g)
        else:
            accum_face = accum_face * 0.95 + face_g * 0.05
            face_low_cut = face_g - accum_face
            
        # 内積積算
        inner_map += face_low_cut * ref_signal[i]
        
    return inner_map

# --- 実行 ---
try:
    # 1. ROIの位置確認
    check_roi_position(UV_MAKEUP_DIR, ROI_X, ROI_Y, ROI_W, ROI_H)

    # 2. UV展開画像から直接信号を取得
    uv_signal = get_signal_from_uv_roi(UV_MAKEUP_DIR, START_F, END_F, ROI_X, ROI_Y, ROI_W, ROI_H)

    # 信号プロット
    plt.figure(figsize=(10, 3))
    plt.plot(uv_signal, color='green', label='Signal from UV ROI')
    plt.title("Reference Signal Extracted from UV Map")
    plt.legend()
    plt.show()

    # 3. 可視化 (内積計算)
    result_map = visualize_correlation(UV_MAKEUP_DIR, uv_signal, START_F)

    # 4. 表示 (従来手法と完全に一致する可視化処理 + 両画面の背景白化)
    # 表示用に最初の画像を読み込む
    files = sorted(list(UV_MAKEUP_DIR.glob("*.npy")))
    first_img = np.load(files[START_F])
    disp_first_img = np.clip(first_img, 0, 255).astype(np.uint8)

    # --- ★背景マスクの作成 ---
    # Gチャンネルが0.0より大きい場所をTrue（顔）とする
    mask = (first_img[:, :, 1] > 0.0)
    
    # --- ★追加: 左側表示用のRGB画像を作り、背景を白(255, 255, 255)で塗りつぶす ---
    disp_first_img_rgb = cv2.cvtColor(disp_first_img, cv2.COLOR_BGR2RGB)
    disp_first_img_rgb[~mask] = [255, 255, 255]
    # -----------------------------------------------------------------------

    fig = plt.figure(figsize=(14, 6))
    grid = ImageGrid(fig, 111, nrows_ncols=(1, 2), axes_pad=0.3,
                     cbar_location="right", cbar_mode="single", cbar_size="5%", cbar_pad=0.1)

    # --- ① 左側：元のRGB画像と赤枠 ---
    # ★変更: 背景を白くしたRGB画像をそのまま渡す
    grid[0].imshow(disp_first_img_rgb)
    
    # 参照信号を取得したROIの赤枠を描画
    rect = plt.Rectangle(
        (ROI_X, ROI_Y),  
        ROI_W,  
        ROI_H,  
        linewidth=2, edgecolor='r', facecolor='none'
    )
    grid[0].add_patch(rect)
    grid[0].set_axis_off()

    # --- ② 右側：可視化画像 ---
    # 1. 顔の概算領域をマウスドラッグで手動取得
    print("顔の領域（明るさ調整用）をドラッグで選択し、SpaceかEnterを押してください。")
    
    # ROI選択ウィンドウを表示 (ここは黒背景のままでも選択操作には影響しません)
    roi_vmax = cv2.selectROI("Select Face Region for Contrast", disp_first_img, showCrosshair=True, fromCenter=False)
    cv2.destroyWindow("Select Face Region for Contrast")
    
    x, y, w, h = roi_vmax
    
    # フェイルセーフ
    if w == 0 or h == 0:
        print("領域が選択されなかったため、デフォルト（中央部分）を使用します。")
        h_img, w_img = result_map.shape
        face_region = result_map[h_img//4 : 3*h_img//4, w_img//4 : 3*w_img//4]
    else:
        face_region = result_map[y : y+h, x : x+w]
    
    # 2. 実質的な最大値（99パーセンタイル）を計算
    face_vmax = np.percentile(face_region, 99)
    if face_vmax <= 0:
        face_vmax = 1.0 
        
    # 3. 正規化
    norm_inner = np.clip(result_map, 0, face_vmax) / face_vmax

    # ==========================================
    # ★追加: 従来内積のマップを保存します
    # ガンマ補正前の、物理的な振幅比（正規化済）を保存します
    np.save("legacy_inner_map.npy", norm_inner)
    print("従来内積のマップを 'legacy_inner_map.npy' として保存しました。")
    # ==========================================
    
    # 4. ガンマ補正
    gamma = 0.5 
    disp_data = norm_inner ** gamma

    # --- ヒートマップの背景白化処理 ---
    disp_data_masked = disp_data.copy()
    disp_data_masked[~mask] = np.nan
    
    cmap_custom = copy.copy(plt.get_cmap('viridis'))
    cmap_custom.set_bad(color='white')
    # --------------------------------
    
    # 表示 (cmapと表示データを変更)
    im = grid[1].imshow(disp_data_masked, cmap=cmap_custom, vmin=0.0, vmax=1.0)
    grid[1].set_axis_off()

    # --- ③ カラーバー ---
    grid.cbar_axes[0].colorbar(im)

    plt.show()

except Exception as e:
    import traceback
    traceback.print_exc()
    print(f"エラー: {e}")
```

```python
#内積キャリブレーション
import numpy as np
import matplotlib.pyplot as plt
import cv2
from pathlib import Path
from tqdm import tqdm
from mpl_toolkits.axes_grid1 import ImageGrid
import copy
from scipy.ndimage import median_filter # ★空間フィルタ用に追加

# --- 設定項目 ---
DIR_BEFORE = Path(r"C:\Users\hil\Documents\cosme\output_uvs_texture\bare\exp004_20260325_14_23_bare1_p70_uvs")
DIR_AFTER  = Path(r"C:\Users\hil\Documents\cosme\output_uvs_texture\fund\exp004_20260325_14_34_fund3_1_p70_uvs")

ROI_X, ROI_Y, ROI_W, ROI_H = 80, 130, 15, 15
START_F, END_F = 0, 300
# ----------------

def get_signal_from_uv_roi(img_dir: Path, start_f, end_f, x, y, w, h):
    """ROIから中央値(Median)を用いて参照信号を抽出"""
    files = sorted(list(img_dir.glob("*.npy")))
    files = files[start_f : min(len(files), end_f)]
    
    signal = []
    accum_val = None
    
    for f_path in tqdm(files, desc=f"Signal Ext ({img_dir.name[:10]}...)"):
        img = np.load(f_path)
        
        # ★修正箇所1: 画像全体(Gチャンネル)に3x3空間メディアンフィルタを適用
        raw_g = img[:, :, 1]
        filtered_g = median_filter(raw_g, size=(3, 3))
        
        # フィルタ後の画像からROIを切り出し
        roi = filtered_g[y:y+h, x:x+w]
        
        # ★修正箇所2: np.mean ではなく np.median を使用
        val = np.median(roi)
        
        if accum_val is None:
            accum_val = val
            ac = 0.0
        else:
            accum_val = accum_val * 0.95 + val * 0.05
            ac = val - accum_val
            
        signal.append(ac)
        
    sig_array = np.array(signal)
    # 参照信号を分散1に正規化（スケールを揃えるための超重要処理）
    if np.std(sig_array) > 0:
        sig_array = (sig_array - np.mean(sig_array)) / np.std(sig_array)
        
    return sig_array

def compute_inner_map(img_dir: Path, ref_signal: np.ndarray, start_f: int):
    """内積マップを計算"""
    files = sorted(list(img_dir.glob("*.npy")))
    files = files[start_f : start_f + len(ref_signal)]
    
    h_map, w_map = 256, 256
    accum_face = np.zeros((h_map, w_map), dtype=np.float64)
    inner_map = np.zeros((h_map, w_map), dtype=np.float64)
    
    for i, f_path in enumerate(tqdm(files, desc=f"Inner Map ({img_dir.name[:10]}...)")):
        img = np.load(f_path)
        
        # ★修正箇所3: マップ計算時にも同様に3x3空間メディアンフィルタを適用
        raw_g = img[:, :, 1]
        filtered_g = median_filter(raw_g, size=(3, 3))
        
        face_g = filtered_g.astype(np.float64)
        
        if i == 0:
            accum_face = face_g
            face_low_cut = np.zeros_like(face_g)
        else:
            accum_face = accum_face * 0.95 + face_g * 0.05
            face_low_cut = face_g - accum_face
            
        inner_map += face_low_cut * ref_signal[i]
        
    return inner_map

def visualize_inner_calibration(inner_before, inner_after, first_img):
    """内積のBefore/After差分を可視化（背景を白でマスク）"""
    
    # --- 1. 完璧な背景マスク（値が0.0の場所を背景とする） ---
    # 先日のgenerate_uvs.pyの修正により、背景は完全に0.0になっている前提
    mask = (first_img[:, :, 1] > 0.0)
    
    # 背景をNaNにして除外
    inner_before_masked = inner_before.copy()
    inner_after_masked = inner_after.copy()
    inner_before_masked[~mask] = np.nan
    inner_after_masked[~mask] = np.nan
    
    # --- 2. 差分計算 ---
    # 振幅の減少を見るため: After - Before (マイナスになるほど厚塗り)
    delta_map = inner_after_masked - inner_before_masked
    
    # ==========================================
    # ★追加: ここで差分マップをファイルとして保存します
    np.save("inner_delta_map.npy", delta_map)
    print("内積の差分マップを 'inner_delta_map.npy' として保存しました。")
    # ==========================================
    # --- 3. カラースケールの最適化 (99パーセンタイルで外れ値除外) ---
    valid_b = inner_before_masked[mask]
    valid_a = inner_after_masked[mask]
    
    if len(valid_b) > 0 and len(valid_a) > 0:
        all_valid = np.concatenate([valid_b, valid_a])
        vmin_val = np.nanpercentile(all_valid, 1)  # ほぼ0
        vmax_val = np.nanpercentile(all_valid, 99) # 実質的な最大振幅
    else:
        vmin_val, vmax_val = 0, 1000
        
    # 差分のスケール（0を中心にするため、絶対値の最大をとる）
    delta_vmax = max(abs(np.nanpercentile(delta_map[mask], 5)), 
                     abs(np.nanpercentile(delta_map[mask], 95)))
    
    # --- 4. 可視化処理 ---
    fig = plt.figure(figsize=(18, 6))
    # grid = ImageGrid(fig, 111, nrows_ncols=(1, 3), axes_pad=0.3,
    #                  cbar_location="right", cbar_mode="each", cbar_size="5%", cbar_pad=0.1)
    grid = ImageGrid(fig, 111, nrows_ncols=(1, 3), axes_pad=0.3)
    
    # 背景を白に
    cmap_custom = copy.copy(plt.get_cmap('viridis'))
    cmap_custom.set_bad(color='white') 
    
    # 1. Before
    im0 = grid[0].imshow(inner_before_masked, cmap=cmap_custom, vmin=max(0, vmin_val), vmax=vmax_val)
    grid[0].set_title("Before Inner Product (Amplitude)")
    grid[0].axis("off")
    #grid.cbar_axes[0].colorbar(im0)

    # 2. After
    im1 = grid[1].imshow(inner_after_masked, cmap=cmap_custom, vmin=max(0, vmin_val), vmax=vmax_val)
    grid[1].set_title("After Inner Product (Amplitude)")
    grid[1].axis("off")
    #grid.cbar_axes[1].colorbar(im1)

    # 3. Difference (After - Before)
    # 差分はマイナスにもなるので、中央が0のNormalizeを使用
    div_norm = plt.Normalize(vmin=-delta_vmax, vmax=delta_vmax) 
    im2 = grid[2].imshow(delta_map, cmap=cmap_custom, norm=div_norm)
    grid[2].set_title("Difference (After - Before)\nPurple: Concealed, Yellow: No Change")
    grid[2].axis("off")
    #grid.cbar_axes[2].colorbar(im2)

    plt.show()

# --- 実行 ---
try:
    print("【1. Before処理】")
    sig_before = get_signal_from_uv_roi(DIR_BEFORE, START_F, END_F, ROI_X, ROI_Y, ROI_W, ROI_H)
    inner_before = compute_inner_map(DIR_BEFORE, sig_before, START_F)
    
    print("\n【2. After処理】")
    sig_after = get_signal_from_uv_roi(DIR_AFTER, START_F, END_F, ROI_X, ROI_Y, ROI_W, ROI_H)
    inner_after = compute_inner_map(DIR_AFTER, sig_after, START_F)
    
    # 最初の画像を取得してマスク用に使用
    first_file = sorted(list(DIR_BEFORE.glob("*.npy")))[START_F]
    first_img = np.load(first_file)
    
    print("\n【3. キャリブレーション結果の可視化】")
    visualize_inner_calibration(inner_before, inner_after, first_img)

except Exception as e:
    import traceback
    traceback.print_exc()
    print(f"エラー: {e}")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from tqdm import tqdm
import mediapipe as mp
from mediapipe.tasks import python
from mediapipe.tasks.python import vision
import os
from scipy.ndimage import median_filter

# --- 設定項目 ---
UV_MAKEUP_DIR = Path(r"C:\Users\hil\Documents\cosme\output_uvs_texture\fund\exp001_20260323_15_16_fund3_1_p70_uvs")
ORIG_MAKEUP_DIR = Path(r"C:\Users\hil\Documents\vimba_rgb_cam\20260323\exp001_20260323_14_35_bare0_p70") 

ROI_X, ROI_Y, ROI_W, ROI_H = 80, 130, 15, 15
START_F, END_F = 0, 300
OBJ_PATH = "canonical_face_model.obj"
# ----------------

# --- 関数定義 ---

def get_signal_from_uv_roi(img_dir: Path, start_f, end_f, x, y, w, h):
    """UV展開画像群の特定領域(ROI)から緑色信号(AC成分)を抽出"""
    files = sorted(list(img_dir.glob("*.npy")))
    files = files[start_f : min(len(files), end_f)]
    
    signal = []
    accum_val = None
    
    print(f"信号抽出中 (UV Map ROI): {img_dir.name}")
    
    for f_path in tqdm(files):
        img = np.load(f_path)
        roi = img[y:y+h, x:x+w, 1]
        val = np.mean(roi)
        
        if accum_val is None:
            accum_val = val
            ac = 0.0
        else:
            accum_val = accum_val * 0.95 + val * 0.05
            ac = val - accum_val
            
        signal.append(ac)
        
    return np.array(signal)

def visualize_correlation(uv_dir: Path, ref_signal: np.ndarray, start_f: int):
    """UV画像群と参照信号の内積マップを作成"""
    files = sorted(list(uv_dir.glob("*.npy")))
    files = files[start_f : start_f + len(ref_signal)]
    
    h_map, w_map = 256, 256
    accum_face = np.zeros((h_map, w_map), dtype=np.float64)
    inner_map = np.zeros((h_map, w_map), dtype=np.float64)
    
    print("内積マップ計算中...")
    
    for i, f_path in enumerate(tqdm(files)):
        img = np.load(f_path)
        face_g = img[:, :, 1].astype(np.float64)
        
        if i == 0:
            accum_face = face_g
            face_low_cut = np.zeros_like(face_g)
        else:
            accum_face = accum_face * 0.95 + face_g * 0.05
            face_low_cut = face_g - accum_face
            
        inner_map += face_low_cut * ref_signal[i]
        
    return inner_map

def warp_triangle(img_dest, img_src, tri_src, tri_dest):
    """三角形ワープ関数"""
    r1 = cv2.boundingRect(np.float32([tri_src]))
    r2 = cv2.boundingRect(np.float32([tri_dest]))
    
    tri_src_cropped = [(t[0]-r1[0], t[1]-r1[1]) for t in tri_src]
    tri_dest_cropped = [(t[0]-r2[0], t[1]-r2[1]) for t in tri_dest]
    
    img_src_cropped = img_src[r1[1]:r1[1]+r1[3], r1[0]:r1[0]+r1[2]]
    
    warp_mat = cv2.getAffineTransform(np.float32(tri_src_cropped), np.float32(tri_dest_cropped))
    img_dest_cropped = cv2.warpAffine(img_src_cropped, warp_mat, (r2[2], r2[3]), 
                                      flags=cv2.INTER_LINEAR, 
                                      borderMode=cv2.BORDER_REFLECT_101)
    
    mask = np.zeros((r2[3], r2[2], 3), dtype=np.float32)
    cv2.fillConvexPoly(mask, np.int32(tri_dest_cropped), (1.0, 1.0, 1.0), 16, 0)
    
    y1, y2 = r2[1], r2[1]+r2[3]
    x1, x2 = r2[0], r2[0]+r2[2]
    
    if y2 > img_dest.shape[0] or x2 > img_dest.shape[1]: return
    
    img_dest_region = img_dest[y1:y2, x1:x2]
    if img_dest_region.shape == img_dest_cropped.shape:
        img_dest[y1:y2, x1:x2] = img_dest_region * (1 - mask) + img_dest_cropped * mask

def load_face_mesh_topology(obj_path):
    """OBJファイルからUV座標とトポロジーを読み込む"""
    if not os.path.exists(obj_path): raise FileNotFoundError(f"{obj_path} が見つかりません。")
    
    with open(obj_path, 'r') as f: lines = f.read().split('\n')
    
    uv_coords = []
    tri_vertex_indices = []
    tri_uv_indices = []
    
    for line in lines:
        if line.startswith('vt '):
            parts = line.split()
            uv_coords.append([float(parts[1]) * 256, float(parts[2]) * 256])
        elif line.startswith('f '):
            parts = line.split()
            v_indices = [int(p.split('/')[0])-1 for p in parts[1:4]]
            tri_vertex_indices.append(v_indices)
            vt_indices = [int(p.split('/')[1])-1 for p in parts[1:4]]
            tri_uv_indices.append(vt_indices)
            
    return np.array(uv_coords), tri_vertex_indices, tri_uv_indices

def rewrap_texture(uv_texture, landmarks, triangles_v, triangles_vt, uv_coords, dest_shape):
    """リラッピング関数"""
    output_img = np.zeros((dest_shape[0], dest_shape[1], 3), dtype=np.uint8)
    
    for i in range(len(triangles_v)):
        tri_v = triangles_v[i]
        tri_vt = triangles_vt[i]
        tri_uv = np.array([uv_coords[idx] for idx in tri_vt], dtype=np.float32)
        tri_img = np.array([landmarks[idx] for idx in tri_v], dtype=np.float32)
        warp_triangle(output_img, uv_texture, tri_uv, tri_img)
        
    return output_img

# --- メイン実行処理 ---

try:
    # 0. 初期化
    script_dir = os.path.dirname(os.path.abspath("__file__"))
    model_path = "face_landmarker.task"
    if not os.path.exists(model_path): model_path = os.path.join(script_dir, "face_landmarker.task")
    with open(model_path, 'rb') as f: model_buffer = f.read()
    base_options = python.BaseOptions(model_asset_buffer=model_buffer)
    detector = vision.FaceLandmarker.create_from_options(vision.FaceLandmarkerOptions(base_options=base_options, num_faces=1))
    
    uv_coords, tri_v_indices, tri_uv_indices = load_face_mesh_topology(OBJ_PATH)

    # 1. 信号抽出 & 2. 内積マップ作成
    uv_signal = get_signal_from_uv_roi(UV_MAKEUP_DIR, START_F, END_F, ROI_X, ROI_Y, ROI_W, ROI_H)
    result_map = visualize_correlation(UV_MAKEUP_DIR, uv_signal, START_F)
    
    # 3. 画像変換 (カラーマップ適用: Matplotlib標準スケール) ★変更点
    gamma = 0.25 # 指定されたガンマ値
    
    # 正規化処理: 最大値で割ることで 0.0 ~ 1.0 の範囲にします
    max_val = np.max(np.abs(result_map))
    if max_val == 0: max_val = 1.0
    
    # 負の値は0にクリップし、最大値で正規化してガンマ補正
    norm_map = (np.clip(result_map, 0, None) / max_val) ** gamma
    
    # カラーマップ適用 (Matplotlibのデフォルト 'viridis' を使用)
    # cmap(norm_map) は 0.0~1.0 の RGBA 配列を返します
    cmap = plt.get_cmap('viridis')
    colored_map_rgba = cmap(norm_map)
    
    # RGBA (0.0-1.0) -> RGB (0-255) に変換
    texture_img_rgb = (colored_map_rgba[:, :, :3] * 255).astype(np.uint8)
    
    # OpenCVで扱うために RGB -> BGR に変換
    original_texture = cv2.cvtColor(texture_img_rgb, cv2.COLOR_RGB2BGR)

    # 4. リラッピング
    orig_files = sorted(list(ORIG_MAKEUP_DIR.glob("*.npy")))
    if not orig_files: raise FileNotFoundError(f"元の画像が見つかりません: {ORIG_MAKEUP_DIR}")
    img_orig = np.load(orig_files[0])
    if img_orig.dtype != np.uint8: img_orig = np.clip(img_orig, 0, 255).astype(np.uint8)
    
    h_orig, w_orig = img_orig.shape[:2]
    mp_img = mp.Image(image_format=mp.ImageFormat.SRGB, data=cv2.cvtColor(img_orig, cv2.COLOR_BGR2RGB))
    detection = detector.detect(mp_img)
    
    rewrapped_face = np.zeros_like(img_orig)
    if detection.face_landmarks:
        lms = detection.face_landmarks[0]
        landmarks_xy = np.array([(lm.x * w_orig, lm.y * h_orig) for lm in lms])
        
        print("顔への再マッピング中 (Color)...")
        # 作成したカラーテクスチャを使用してリラッピング
        rewrapped_face = rewrap_texture(original_texture, landmarks_xy, tri_v_indices, tri_uv_indices, uv_coords, (h_orig, w_orig))

    # 5. 合成
    blended_face = img_orig.copy()
    mask = (np.sum(rewrapped_face, axis=2) > 0).astype(np.uint8)
    mask_3ch = cv2.merge([mask, mask, mask])
    
    overlay = cv2.addWeighted(img_orig, 0.0, rewrapped_face, 1.0, 0)
    np.copyto(blended_face, overlay, where=(mask_3ch > 0))

    # 6. プロット
    plt.figure(figsize=(12, 6))
    
    # 左：UVマップ (BGRで作ったのでRGBに戻して表示)
    plt.subplot(1, 2, 1)
    plt.imshow(cv2.cvtColor(original_texture, cv2.COLOR_BGR2RGB)) 
    plt.title("1. UV Map (Viridis Scale)")
    plt.axis('off')
    
    # 右：合成結果 (BGRで作ったのでRGBに戻して表示)
    plt.subplot(1, 2, 2)
    plt.imshow(cv2.cvtColor(blended_face, cv2.COLOR_BGR2RGB))
    plt.title("2. Rewrapped on Original Face")
    plt.axis('off')
    plt.show()

except Exception as e:
    import traceback
    traceback.print_exc()
    print(f"エラー: {e}")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from tqdm import tqdm
import mediapipe as mp
from mediapipe.tasks import python
from mediapipe.tasks.python import vision
import os

# --- 設定項目 ---
# 1. 元の .npy ファイル（化粧時）が入っているディレクトリ
# ※ generate_uvs.py で指定した MAKEUP_DIR と同じパスにしてください
ORIGINAL_MAKEUP_DIR = Path(r"E:\cosme_subjects\20250110cosme\exp005_20250110_14_31_fundA1_p85")

# 2. UV展開後の .npy ファイルが入っているディレクトリ
UV_MAKEUP_DIR = Path("../output_uv_textures/makeup")

# 解析フレーム数
START_F = 0
END_F = 300
# ----------------

def initialize_landmarker():
    """MediaPipe FaceLandmarkerの初期化"""
    script_dir = os.path.dirname(os.path.abspath("__file__")) # Notebookならカレント
    # カレントまたは同階層を探す
    path_candidates = [
        "face_landmarker.task",
        os.path.join(script_dir, "face_landmarker.task")
    ]
    model_path = next((p for p in path_candidates if os.path.exists(p)), None)
    
    if model_path is None:
        raise FileNotFoundError("face_landmarker.task が見つかりません。")

    with open(model_path, 'rb') as f:
        model_buffer = f.read()

    base_options = python.BaseOptions(model_asset_buffer=model_buffer)
    options = vision.FaceLandmarkerOptions(base_options=base_options, num_faces=1)
    return vision.FaceLandmarker.create_from_options(options)

def get_reference_signal_from_original(img_dir: Path, start_f: int, end_f: int):
    """
    元の画像群から顔（特定部位）を検出し、緑色信号(AC成分)を抽出する
    """
    files = sorted(list(img_dir.glob("*.npy")))
    if not files:
        raise FileNotFoundError(f"元の画像が見つかりません: {img_dir}")
    
    files = files[start_f : min(len(files), end_f)]
    
    detector = initialize_landmarker()
    signal = []
    accum_val = None
    
    # 追跡するランドマークID (101: 右頬/目の下あたり, 10: 額)
    # 安定して脈が取れる場所として 101 を採用します
    TARGET_ID = 101 
    
    print(f"参照信号を抽出中 (Original): {img_dir.name}")
    
    for f_path in tqdm(files):
        # 画像読み込み
        img_bgr = np.load(f_path)
        
        # 検出用にRGB変換 & uint8化
        if img_bgr.dtype != np.uint8:
            img_for_det = np.clip(img_bgr, 0, 255).astype(np.uint8)
        else:
            img_for_det = img_bgr
            
        img_rgb = cv2.cvtColor(img_for_det, cv2.COLOR_BGR2RGB)
        mp_image = mp.Image(image_format=mp.ImageFormat.SRGB, data=img_rgb)
        
        # 顔検出
        detection = detector.detect(mp_image)
        
        val = 0.0
        if detection.face_landmarks:
            lm = detection.face_landmarks[0][TARGET_ID]
            h, w = img_bgr.shape[:2]
            cx, cy = int(lm.x * w), int(lm.y * h)
            
            # ROI切り出し (20x20)
            x1, y1 = max(0, cx - 10), max(0, cy - 10)
            x2, y2 = min(w, cx + 10), min(h, cy + 10)
            
            roi = img_bgr[y1:y2, x1:x2, 1] # Gチャンネル
            if roi.size > 0:
                val = np.mean(roi)
        
        # フィルタ処理 (AC成分抽出)
        if accum_val is None:
            accum_val = val
            ac = 0.0
        else:
            if val == 0.0: # 検出失敗時は直前の値を維持するなどの処理
                val = accum_val 
            
            accum_val = accum_val * 0.95 + val * 0.05
            ac = val - accum_val
            
        signal.append(ac)
        
    return np.array(signal)

def visualize_with_self_signal(uv_dir: Path, ref_signal: np.ndarray, start_f: int):
    """
    UV展開画像群と参照信号の内積を計算して可視化
    """
    files = sorted(list(uv_dir.glob("*.npy")))
    files = files[start_f : start_f + len(ref_signal)]
    
    h, w = 256, 256 # UVマップのサイズ
    accum_face = np.zeros((h, w), dtype=np.float64)
    inner_map = np.zeros((h, w), dtype=np.float64)
    
    print(f"UVマップと同期計算中: {uv_dir.name}")

    for i, f_path in enumerate(tqdm(files)):
        img = np.load(f_path)
        
        # Gチャンネルを取得
        face_g = img[:, :, 1].astype(np.float64)
        
        if i == 0:
            accum_face = face_g
            face_low_cut = np.zeros_like(face_g)
        else:
            # フィルタ処理
            accum_face = accum_face * 0.95 + face_g * 0.05
            face_low_cut = face_g - accum_face
            
        # 内積計算: (各画素の変動) × (参照信号の値)
        inner_map += face_low_cut * ref_signal[i]
        
    return inner_map

# --- メイン処理 ---

try:
    # 1. 元画像から参照信号を取得
    sample_signal = get_reference_signal_from_original(ORIGINAL_MAKEUP_DIR, START_F, END_F)

    # 信号確認用プロット
    plt.figure(figsize=(10, 3))
    plt.plot(sample_signal, color='green', label='Original Signal (Cheek)')
    plt.title("Reference Signal from Original Makeup Images")
    plt.legend()
    plt.show()

    # 2. UV展開画像を使って可視化 (自己相関的なマッピング)
    result_map = visualize_with_self_signal(UV_MAKEUP_DIR, sample_signal, START_F)

    # 3. 表示
    plt.figure(figsize=(6, 6))
    gamma = 0.5
    # 正規化して表示
    disp_img = (np.clip(result_map, 0, None) / (np.max(np.abs(result_map)) + 1e-6)) ** gamma
    
    plt.imshow(disp_img, cmap='gray', vmin=0)
    plt.title("Visualization on UV Map\n(Ref: Original Image Signal)")
    plt.axis('off')
    plt.show()

except Exception as e:
    print(f"エラー: {e}")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from tqdm import tqdm
import mediapipe as mp
from mediapipe.tasks import python
from mediapipe.tasks.python import vision
import os

# --- 設定項目 ---
UV_MAKEUP_DIR = Path(r"C:\Users\hil\Documents\cosme\output_uvs_texture\fund\exp001_20260107_14_25_fund1_p75_uvs")
ORIG_MAKEUP_DIR = Path(r"E:\20260107\exp001_20260107_14_25_fund1_p75") 

ROI_X, ROI_Y, ROI_W, ROI_H = 80, 130, 15, 15
START_F, END_F = 0, 300
OBJ_PATH = "canonical_face_model.obj"
# ----------------

# --- 関数定義 ---

def get_signal_from_uv_roi(img_dir: Path, start_f, end_f, x, y, w, h):
    """UV展開画像群の特定領域(ROI)から緑色信号(AC成分)を抽出"""
    files = sorted(list(img_dir.glob("*.npy")))
    files = files[start_f : min(len(files), end_f)]
    
    signal = []
    accum_val = None
    
    print(f"信号抽出中 (UV Map ROI): {img_dir.name}")
    
    for f_path in tqdm(files):
        img = np.load(f_path)
        roi = img[y:y+h, x:x+w, 1]
        val = np.mean(roi)
        
        if accum_val is None:
            accum_val = val
            ac = 0.0
        else:
            accum_val = accum_val * 0.95 + val * 0.05
            ac = val - accum_val
            
        signal.append(ac)
        
    return np.array(signal)

def visualize_correlation(uv_dir: Path, ref_signal: np.ndarray, start_f: int):
    """UV画像群と参照信号の内積マップを作成"""
    files = sorted(list(uv_dir.glob("*.npy")))
    files = files[start_f : start_f + len(ref_signal)]
    
    h_map, w_map = 256, 256
    accum_face = np.zeros((h_map, w_map), dtype=np.float64)
    inner_map = np.zeros((h_map, w_map), dtype=np.float64)
    
    print("内積マップ計算中...")
    
    for i, f_path in enumerate(tqdm(files)):
        img = np.load(f_path)
        face_g = img[:, :, 1].astype(np.float64)
        
        if i == 0:
            accum_face = face_g
            face_low_cut = np.zeros_like(face_g)
        else:
            accum_face = accum_face * 0.95 + face_g * 0.05
            face_low_cut = face_g - accum_face
            
        inner_map += face_low_cut * ref_signal[i]
        
    return inner_map

def warp_triangle(img_dest, img_src, tri_src, tri_dest):
    """三角形ワープ関数"""
    r1 = cv2.boundingRect(np.float32([tri_src]))
    r2 = cv2.boundingRect(np.float32([tri_dest]))
    
    tri_src_cropped = [(t[0]-r1[0], t[1]-r1[1]) for t in tri_src]
    tri_dest_cropped = [(t[0]-r2[0], t[1]-r2[1]) for t in tri_dest]
    
    img_src_cropped = img_src[r1[1]:r1[1]+r1[3], r1[0]:r1[0]+r1[2]]
    
    warp_mat = cv2.getAffineTransform(np.float32(tri_src_cropped), np.float32(tri_dest_cropped))
    img_dest_cropped = cv2.warpAffine(img_src_cropped, warp_mat, (r2[2], r2[3]), 
                                      flags=cv2.INTER_LINEAR, 
                                      borderMode=cv2.BORDER_REFLECT_101)
    
    mask = np.zeros((r2[3], r2[2], 3), dtype=np.float32)
    cv2.fillConvexPoly(mask, np.int32(tri_dest_cropped), (1.0, 1.0, 1.0), 16, 0)
    
    y1, y2 = r2[1], r2[1]+r2[3]
    x1, x2 = r2[0], r2[0]+r2[2]
    
    if y2 > img_dest.shape[0] or x2 > img_dest.shape[1]: return
    
    img_dest_region = img_dest[y1:y2, x1:x2]
    if img_dest_region.shape == img_dest_cropped.shape:
        img_dest[y1:y2, x1:x2] = img_dest_region * (1 - mask) + img_dest_cropped * mask

def load_face_mesh_topology(obj_path):
    """
    OBJファイルからUV座標とトポロジーを読み込む（修正版）
    頂点インデックスとUVインデックスを別々に取得します。
    """
    if not os.path.exists(obj_path): raise FileNotFoundError(f"{obj_path} が見つかりません。")
    
    with open(obj_path, 'r') as f: lines = f.read().split('\n')
    
    uv_coords = []
    tri_vertex_indices = [] # 形状用のインデックス
    tri_uv_indices = []     # テクスチャ用のインデックス
    
    for line in lines:
        if line.startswith('vt '):
            parts = line.split()
            # UV座標の読み込み
            uv_coords.append([float(parts[1]) * 256, float(parts[2]) * 256])
            
        elif line.startswith('f '):
            parts = line.split()
            # f v1/vt1/vn1 v2/vt2/vn2 v3/vt3/vn3
            
            # 頂点インデックス (スラッシュの1つ目)
            v_indices = [int(p.split('/')[0])-1 for p in parts[1:4]]
            tri_vertex_indices.append(v_indices)
            
            # UVインデックス (スラッシュの2つ目)
            vt_indices = [int(p.split('/')[1])-1 for p in parts[1:4]]
            tri_uv_indices.append(vt_indices)
            
    return np.array(uv_coords), tri_vertex_indices, tri_uv_indices

def rewrap_texture(uv_texture, landmarks, triangles_v, triangles_vt, uv_coords, dest_shape):
    """
    リラッピング関数（修正版）
    頂点インデックスとUVインデックスを使い分けます。
    """
    output_img = np.zeros((dest_shape[0], dest_shape[1], 3), dtype=np.uint8)
    
    # triangles_v: 頂点のつなぎ方, triangles_vt: UVのつなぎ方
    for i in range(len(triangles_v)):
        tri_v = triangles_v[i]   # [v1, v2, v3]
        tri_vt = triangles_vt[i] # [vt1, vt2, vt3]
        
        # UV座標は UVインデックス(vt) を使って取得
        tri_uv = np.array([uv_coords[idx] for idx in tri_vt], dtype=np.float32)
        
        # 画像座標は 頂点インデックス(v) を使って取得
        tri_img = np.array([landmarks[idx] for idx in tri_v], dtype=np.float32)
        
        warp_triangle(output_img, uv_texture, tri_uv, tri_img)
        
    return output_img

# --- メイン実行処理 ---

try:
    # 0. 初期化
    script_dir = os.path.dirname(os.path.abspath("__file__"))
    model_path = "face_landmarker.task"
    if not os.path.exists(model_path): model_path = os.path.join(script_dir, "face_landmarker.task")
    with open(model_path, 'rb') as f: model_buffer = f.read()
    base_options = python.BaseOptions(model_asset_buffer=model_buffer)
    detector = vision.FaceLandmarker.create_from_options(vision.FaceLandmarkerOptions(base_options=base_options, num_faces=1))
    
    # 修正版関数で読み込み
    uv_coords, tri_v_indices, tri_uv_indices = load_face_mesh_topology(OBJ_PATH)

    # 1. 信号抽出 & 2. 内積マップ作成
    uv_signal = get_signal_from_uv_roi(UV_MAKEUP_DIR, START_F, END_F, ROI_X, ROI_Y, ROI_W, ROI_H)
    result_map = visualize_correlation(UV_MAKEUP_DIR, uv_signal, START_F)
    
    # 3. 画像変換 (グレースケール)
    gamma = 0.5
    norm_map = (np.clip(result_map, 0, None) / (np.max(np.abs(result_map)) + 1e-6)) ** gamma
    gray_map = (norm_map * 255).astype(np.uint8)
    original_texture = cv2.merge([gray_map, gray_map, gray_map])

    # 4. リラッピング
    orig_files = sorted(list(ORIG_MAKEUP_DIR.glob("*.npy")))
    if not orig_files: raise FileNotFoundError(f"元の画像が見つかりません: {ORIG_MAKEUP_DIR}")
    img_orig = np.load(orig_files[0])
    if img_orig.dtype != np.uint8: img_orig = np.clip(img_orig, 0, 255).astype(np.uint8)
    
    h_orig, w_orig = img_orig.shape[:2]
    mp_img = mp.Image(image_format=mp.ImageFormat.SRGB, data=cv2.cvtColor(img_orig, cv2.COLOR_BGR2RGB))
    detection = detector.detect(mp_img)
    
    rewrapped_face = np.zeros_like(img_orig)
    if detection.face_landmarks:
        lms = detection.face_landmarks[0]
        landmarks_xy = np.array([(lm.x * w_orig, lm.y * h_orig) for lm in lms])
        
        print("顔への再マッピング中 (修正版: 正しいUVインデックスを使用)...")
        # 引数に tri_v_indices と tri_uv_indices の両方を渡す
        rewrapped_face = rewrap_texture(original_texture, landmarks_xy, tri_v_indices, tri_uv_indices, uv_coords, (h_orig, w_orig))

    # 5. 合成
    blended_face = img_orig.copy()
    mask = (np.sum(rewrapped_face, axis=2) > 0).astype(np.uint8)
    mask_3ch = cv2.merge([mask, mask, mask])
    
    overlay = cv2.addWeighted(img_orig, 0.0, rewrapped_face, 1.0, 0)
    np.copyto(blended_face, overlay, where=(mask_3ch > 0))

    # 6. プロット
    plt.figure(figsize=(12, 6))
    plt.subplot(1, 2, 1)
    plt.imshow(original_texture, cmap='gray', vmin=0) 
    plt.title("1. UV Map")
    plt.axis('off')
    
    plt.subplot(1, 2, 2)
    plt.imshow(cv2.cvtColor(blended_face, cv2.COLOR_BGR2RGB))
    plt.title("2. Rewrapped on Original Face")
    plt.axis('off')
    plt.show()

except Exception as e:
    import traceback
    traceback.print_exc()
    print(f"エラー: {e}")
```

```python
import cv2
import numpy as np
from pathlib import Path
from tqdm import tqdm
import mediapipe as mp
from mediapipe.tasks import python
from mediapipe.tasks.python import vision
import os
import matplotlib.pyplot as plt

# --- 設定項目 ---
UV_MAKEUP_DIR = Path("../output_uv_textures/makeup")
ORIG_MAKEUP_DIR = Path(r"E:\cosme_subjects\20250110cosme\exp005_20250110_14_31_fundA1_p85") 

ROI_X, ROI_Y, ROI_W, ROI_H = 80, 130, 15, 15
START_F, END_F = 0, 300
OBJ_PATH = "canonical_face_model.obj"

# ★ ディスプレイ設定
SCREEN_W = 1920
SCREEN_H = 1080
DISPLAY_OFFSET_X = 1920 
DISPLAY_OFFSET_Y = 0

# ★ 表示サイズの調整 (1.0 = 画面フィット, 0.5 = 半分)
DISPLAY_SCALE = 0.7  # ここを小さくすると顔が小さくなります
# ----------------

# --- 関数定義 ---
def get_signal_from_uv_roi(img_dir: Path, start_f, end_f, x, y, w, h):
    files = sorted(list(img_dir.glob("*.npy")))
    files = files[start_f : min(len(files), end_f)]
    signal = []
    accum_val = None
    print(f"信号抽出中: {img_dir.name}")
    for f_path in tqdm(files):
        img = np.load(f_path)
        roi = img[y:y+h, x:x+w, 1]
        val = np.mean(roi)
        if accum_val is None: accum_val = val; ac = 0.0
        else: accum_val = accum_val * 0.95 + val * 0.05; ac = val - accum_val
        signal.append(ac)
    return np.array(signal)

def visualize_correlation(uv_dir: Path, ref_signal: np.ndarray, start_f: int):
    files = sorted(list(uv_dir.glob("*.npy")))
    files = files[start_f : start_f + len(ref_signal)]
    h_map, w_map = 256, 256
    accum_face = np.zeros((h_map, w_map), dtype=np.float64)
    inner_map = np.zeros((h_map, w_map), dtype=np.float64)
    print("内積マップ計算中...")
    for i, f_path in enumerate(tqdm(files)):
        img = np.load(f_path)
        face_g = img[:, :, 1].astype(np.float64)
        if i == 0: accum_face = face_g; face_low_cut = np.zeros_like(face_g)
        else: accum_face = accum_face * 0.95 + face_g * 0.05; face_low_cut = face_g - accum_face
        inner_map += face_low_cut * ref_signal[i]
    return inner_map

def warp_triangle(img_dest, img_src, tri_src, tri_dest):
    r1 = cv2.boundingRect(np.float32([tri_src]))
    r2 = cv2.boundingRect(np.float32([tri_dest]))
    tri_src_cropped = [(t[0]-r1[0], t[1]-r1[1]) for t in tri_src]
    tri_dest_cropped = [(t[0]-r2[0], t[1]-r2[1]) for t in tri_dest]
    img_src_cropped = img_src[r1[1]:r1[1]+r1[3], r1[0]:r1[0]+r1[2]]
    warp_mat = cv2.getAffineTransform(np.float32(tri_src_cropped), np.float32(tri_dest_cropped))
    img_dest_cropped = cv2.warpAffine(img_src_cropped, warp_mat, (r2[2], r2[3]), flags=cv2.INTER_LINEAR, borderMode=cv2.BORDER_REFLECT_101)
    mask = np.zeros((r2[3], r2[2], 3), dtype=np.float32)
    cv2.fillConvexPoly(mask, np.int32(tri_dest_cropped), (1.0, 1.0, 1.0), 16, 0)
    y1, y2 = r2[1], r2[1]+r2[3]
    x1, x2 = r2[0], r2[0]+r2[2]
    if y2 > img_dest.shape[0] or x2 > img_dest.shape[1]: return
    img_dest_region = img_dest[y1:y2, x1:x2]
    if img_dest_region.shape == img_dest_cropped.shape:
        img_dest[y1:y2, x1:x2] = img_dest_region * (1 - mask) + img_dest_cropped * mask

def load_face_mesh_topology(obj_path):
    if not os.path.exists(obj_path): raise FileNotFoundError(f"{obj_path} が見つかりません。")
    with open(obj_path, 'r') as f: lines = f.read().split('\n')
    uv_coords = []
    tri_vertex_indices = []
    tri_uv_indices = []
    for line in lines:
        if line.startswith('vt '):
            parts = line.split()
            uv_coords.append([float(parts[1]) * 256, float(parts[2]) * 256])
        elif line.startswith('f '):
            parts = line.split()
            v_indices = [int(p.split('/')[0])-1 for p in parts[1:4]]
            tri_vertex_indices.append(v_indices)
            vt_indices = [int(p.split('/')[1])-1 for p in parts[1:4]]
            tri_uv_indices.append(vt_indices)
    return np.array(uv_coords), tri_vertex_indices, tri_uv_indices

def rewrap_texture(uv_texture, landmarks, triangles_v, triangles_vt, uv_coords, dest_shape):
    output_img = np.zeros((dest_shape[0], dest_shape[1], 3), dtype=np.uint8)
    for i in range(len(triangles_v)):
        tri_v = triangles_v[i]
        tri_vt = triangles_vt[i]
        tri_uv = np.array([uv_coords[idx] for idx in tri_vt], dtype=np.float32)
        tri_img = np.array([landmarks[idx] for idx in tri_v], dtype=np.float32)
        warp_triangle(output_img, uv_texture, tri_uv, tri_img)
    return output_img

def draw_face_contour(img):
    """黒背景の画像から非黒領域の輪郭を抽出し、線を引く"""
    img_with_contour = img.copy()
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    _, mask = cv2.threshold(gray, 1, 255, cv2.THRESH_BINARY)
    contours, _ = cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    cv2.drawContours(img_with_contour, contours, -1, (255, 255, 0), 3)
    return img_with_contour

# ★修正: スケール調整機能を追加
def fit_image_to_screen(img, screen_w, screen_h, scale_factor=1.0):
    """画像を画面中央に配置する（スケール調整付き）"""
    h, w = img.shape[:2]
    # 画面に収まる最大スケールに、さらに scale_factor を掛ける
    max_scale = min(screen_w / w, screen_h / h)
    final_scale = max_scale * scale_factor
    
    new_w, new_h = int(w * final_scale), int(h * final_scale)
    
    # サイズが0以下にならないようガード
    if new_w <= 0 or new_h <= 0: return np.zeros((screen_h, screen_w, 3), dtype=np.uint8)
    
    resized_img = cv2.resize(img, (new_w, new_h))
    canvas = np.zeros((screen_h, screen_w, 3), dtype=np.uint8)
    
    x_offset = (screen_w - new_w) // 2
    y_offset = (screen_h - new_h) // 2
    
    canvas[y_offset:y_offset+new_h, x_offset:x_offset+new_w] = resized_img
    return canvas

# --- メイン実行処理 ---

try:
    # 0. 初期化
    script_dir = os.path.dirname(os.path.abspath("__file__"))
    model_path = "face_landmarker.task"
    if not os.path.exists(model_path): model_path = os.path.join(script_dir, "face_landmarker.task")
    with open(model_path, 'rb') as f: model_buffer = f.read()
    base_options = python.BaseOptions(model_asset_buffer=model_buffer)
    detector = vision.FaceLandmarker.create_from_options(vision.FaceLandmarkerOptions(base_options=base_options, num_faces=1))
    
    uv_coords, tri_v_indices, tri_uv_indices = load_face_mesh_topology(OBJ_PATH)

    # 1. 信号抽出 & 2. 内積マップ作成
    uv_signal = get_signal_from_uv_roi(UV_MAKEUP_DIR, START_F, END_F, ROI_X, ROI_Y, ROI_W, ROI_H)
    result_map = visualize_correlation(UV_MAKEUP_DIR, uv_signal, START_F)
    
    # 3. カラーマップ作成 (Viridis)
    gamma = 0.25
    clipped_map = np.clip(result_map, 0, None)
    max_val = np.max(clipped_map)
    if max_val == 0: max_val = 1.0
    norm_map = (clipped_map / max_val) ** gamma
    
    cmap = plt.get_cmap('viridis')
    colored_map_rgba = cmap(norm_map)
    texture_img_bgr = (colored_map_rgba[:, :, :3] * 255).astype(np.uint8)
    texture_img_bgr = cv2.cvtColor(texture_img_bgr, cv2.COLOR_RGB2BGR)

    # 4. リラッピング
    orig_files = sorted(list(ORIG_MAKEUP_DIR.glob("*.npy")))
    if not orig_files: raise FileNotFoundError(f"元の画像が見つかりません: {ORIG_MAKEUP_DIR}")
    img_orig = np.load(orig_files[0])
    if img_orig.dtype != np.uint8: img_orig = np.clip(img_orig, 0, 255).astype(np.uint8)
    
    h_orig, w_orig = img_orig.shape[:2]
    mp_img = mp.Image(image_format=mp.ImageFormat.SRGB, data=cv2.cvtColor(img_orig, cv2.COLOR_BGR2RGB))
    detection = detector.detect(mp_img)
    
    rewrapped_face = np.zeros_like(img_orig)

    if detection.face_landmarks:
        lms = detection.face_landmarks[0]
        landmarks_xy = np.array([(lm.x * w_orig, lm.y * h_orig) for lm in lms])
        
        print("顔への再マッピング中 (カラー)...")
        rewrapped_face = rewrap_texture(texture_img_bgr, landmarks_xy, tri_v_indices, tri_uv_indices, uv_coords, (h_orig, w_orig))
        
        # 輪郭線描画
        print("輪郭線を描画中...")
        rewrapped_face = draw_face_contour(rewrapped_face)

    # 5. ウィンドウ表示
    window_name = "Smart Mirror Display"
    
    # ★サイズ調整を適用
    display_img = fit_image_to_screen(rewrapped_face, SCREEN_W, SCREEN_H, DISPLAY_SCALE)
    
    cv2.namedWindow(window_name, cv2.WINDOW_NORMAL)
    cv2.moveWindow(window_name, DISPLAY_OFFSET_X, DISPLAY_OFFSET_Y)
    cv2.setWindowProperty(window_name, cv2.WND_PROP_FULLSCREEN, cv2.WINDOW_FULLSCREEN)
    
    cv2.imshow(window_name, display_img)
    
    print(f"ウィンドウを表示しました。(オフセット: X={DISPLAY_OFFSET_X}, Y={DISPLAY_OFFSET_Y}, スケール: {DISPLAY_SCALE})")
    print("「q」キーを押すと終了します。")
    
    while True:
        key = cv2.waitKey(1) & 0xFF
        if key == ord('q'):
            break
            
    cv2.destroyAllWindows()

except Exception as e:
    import traceback
    traceback.print_exc()
    print(f"エラー: {e}")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from tqdm import tqdm
import mediapipe as mp
from mediapipe.tasks import python
from mediapipe.tasks.python import vision
import os

# --- 設定項目 ---
# 以前の実行で保存した_uvsフォルダを指定
UV_MAKEUP_DIR = Path("../output_uv_textures/makeup")
# 元の動画（.npy）があるフォルダを指定
ORIG_MAKEUP_DIR = Path(r"E:\cosme_subjects\20250110cosme\exp005_20250110_14_31_fundA1_p85") 

ROI_X, ROI_Y, ROI_W, ROI_H = 80, 130, 15, 15
START_F, END_F = 0, 300
OBJ_PATH = "canonical_face_model.obj"
# ----------------

# --- 関数定義 ---

def get_signal_from_uv_roi(img_dir: Path, start_f, end_f, x, y, w, h):
    files = sorted(list(img_dir.glob("*.npy")))
    files = files[start_f : min(len(files), end_f)]
    signal = []
    accum_val = None
    print(f"信号抽出中 (UV Map ROI): {img_dir.name}")
    for f_path in tqdm(files):
        img = np.load(f_path)
        roi = img[y:y+h, x:x+w, 1]
        val = np.mean(roi)
        if accum_val is None: accum_val = val; ac = 0.0
        else: accum_val = accum_val * 0.95 + val * 0.05; ac = val - accum_val
        signal.append(ac)
    return np.array(signal)

def visualize_correlation(uv_dir: Path, ref_signal: np.ndarray, start_f: int):
    files = sorted(list(uv_dir.glob("*.npy")))
    files = files[start_f : start_f + len(ref_signal)]
    h_map, w_map = 256, 256
    accum_face = np.zeros((h_map, w_map), dtype=np.float64)
    inner_map = np.zeros((h_map, w_map), dtype=np.float64)
    print("内積マップ計算中...")
    for i, f_path in enumerate(tqdm(files)):
        img = np.load(f_path)
        face_g = img[:, :, 1].astype(np.float64)
        if i == 0: accum_face = face_g; face_low_cut = np.zeros_like(face_g)
        else: accum_face = accum_face * 0.95 + face_g * 0.05; face_low_cut = face_g - accum_face
        inner_map += face_low_cut * ref_signal[i]
    return inner_map

def warp_triangle(img_dest, img_src, tri_src, tri_dest):
    r1 = cv2.boundingRect(np.float32([tri_src]))
    r2 = cv2.boundingRect(np.float32([tri_dest]))
    tri_src_cropped = [(t[0]-r1[0], t[1]-r1[1]) for t in tri_src]
    tri_dest_cropped = [(t[0]-r2[0], t[1]-r2[1]) for t in tri_dest]
    img_src_cropped = img_src[r1[1]:r1[1]+r1[3], r1[0]:r1[0]+r1[2]]
    warp_mat = cv2.getAffineTransform(np.float32(tri_src_cropped), np.float32(tri_dest_cropped))
    img_dest_cropped = cv2.warpAffine(img_src_cropped, warp_mat, (r2[2], r2[3]), flags=cv2.INTER_LINEAR, borderMode=cv2.BORDER_REFLECT_101)
    mask = np.zeros((r2[3], r2[2], 3), dtype=np.float32)
    cv2.fillConvexPoly(mask, np.int32(tri_dest_cropped), (1.0, 1.0, 1.0), 16, 0)
    y1, y2 = r2[1], r2[1]+r2[3]
    x1, x2 = r2[0], r2[0]+r2[2]
    if y2 > img_dest.shape[0] or x2 > img_dest.shape[1]: return
    img_dest_region = img_dest[y1:y2, x1:x2]
    if img_dest_region.shape == img_dest_cropped.shape:
        img_dest[y1:y2, x1:x2] = img_dest_region * (1 - mask) + img_dest_cropped * mask

def load_face_mesh_topology(obj_path):
    if not os.path.exists(obj_path): raise FileNotFoundError(f"{obj_path} が見つかりません。")
    with open(obj_path, 'r') as f: lines = f.read().split('\n')
    uv_coords = []
    tri_vertex_indices = []
    tri_uv_indices = []
    for line in lines:
        if line.startswith('vt '):
            parts = line.split()
            uv_coords.append([float(parts[1]) * 256, float(parts[2]) * 256])
        elif line.startswith('f '):
            parts = line.split()
            v_indices = [int(p.split('/')[0])-1 for p in parts[1:4]]
            tri_vertex_indices.append(v_indices)
            vt_indices = [int(p.split('/')[1])-1 for p in parts[1:4]]
            tri_uv_indices.append(vt_indices)
    return np.array(uv_coords), tri_vertex_indices, tri_uv_indices

def rewrap_texture(uv_texture, landmarks, triangles_v, triangles_vt, uv_coords, dest_shape):
    output_img = np.zeros((dest_shape[0], dest_shape[1], 3), dtype=np.uint8)
    for i in range(len(triangles_v)):
        tri_v = triangles_v[i]
        tri_vt = triangles_vt[i]
        tri_uv = np.array([uv_coords[idx] for idx in tri_vt], dtype=np.float32)
        tri_img = np.array([landmarks[idx] for idx in tri_v], dtype=np.float32)
        warp_triangle(output_img, uv_texture, tri_uv, tri_img)
    return output_img

# --- メイン実行処理 ---

try:
    # 0. 初期化
    script_dir = os.path.dirname(os.path.abspath("__file__"))
    model_path = "face_landmarker.task"
    if not os.path.exists(model_path): model_path = os.path.join(script_dir, "face_landmarker.task")
    with open(model_path, 'rb') as f: model_buffer = f.read()
    base_options = python.BaseOptions(model_asset_buffer=model_buffer)
    detector = vision.FaceLandmarker.create_from_options(vision.FaceLandmarkerOptions(base_options=base_options, num_faces=1))
    
    uv_coords, tri_v_indices, tri_uv_indices = load_face_mesh_topology(OBJ_PATH)

    # 1. 信号抽出 & 2. 内積マップ作成
    uv_signal = get_signal_from_uv_roi(UV_MAKEUP_DIR, START_F, END_F, ROI_X, ROI_Y, ROI_W, ROI_H)
    result_map = visualize_correlation(UV_MAKEUP_DIR, uv_signal, START_F)
    
    # 3. 画像変換 (グレースケール)
    gamma = 0.5
    norm_map = (np.clip(result_map, 0, None) / (np.max(np.abs(result_map)) + 1e-6)) ** gamma
    gray_map = (norm_map * 255).astype(np.uint8)
    original_texture = cv2.merge([gray_map, gray_map, gray_map])

    # 4. リラッピング
    orig_files = sorted(list(ORIG_MAKEUP_DIR.glob("*.npy")))
    if not orig_files: raise FileNotFoundError(f"元の画像が見つかりません: {ORIG_MAKEUP_DIR}")
    img_orig = np.load(orig_files[0])
    if img_orig.dtype != np.uint8: img_orig = np.clip(img_orig, 0, 255).astype(np.uint8)
    
    h_orig, w_orig = img_orig.shape[:2]
    mp_img = mp.Image(image_format=mp.ImageFormat.SRGB, data=cv2.cvtColor(img_orig, cv2.COLOR_BGR2RGB))
    detection = detector.detect(mp_img)
    
    # 黒背景用の画像を初期化
    rewrapped_face = np.zeros_like(img_orig)

    if detection.face_landmarks:
        lms = detection.face_landmarks[0]
        landmarks_xy = np.array([(lm.x * w_orig, lm.y * h_orig) for lm in lms])
        
        print("顔への再マッピング中...")
        rewrapped_face = rewrap_texture(original_texture, landmarks_xy, tri_v_indices, tri_uv_indices, uv_coords, (h_orig, w_orig))

    # 5. 表示 (スマートミラー用)
    # 枠や軸を消し、黒背景画像のみを表示
    plt.figure(figsize=(10, 10)) # サイズはお好みで調整してください
    plt.imshow(cv2.cvtColor(rewrapped_face, cv2.COLOR_BGR2RGB))
    
    # 余計な情報を削除
    plt.axis('off')
    
    # Matplotlibの背景色も黒にする（隙間対策）
    plt.gca().set_facecolor('black')
    plt.gcf().patch.set_facecolor('black')
    
    # タイトルも不要なら削除してください
    # plt.title("Smart Mirror View", color='white') 
    
    plt.show()

except Exception as e:
    import traceback
    traceback.print_exc()
    print(f"エラー: {e}")
```

```python
#頂点カラーサンプリング法
import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from tqdm import tqdm
import mediapipe as mp
from mediapipe.tasks import python
from mediapipe.tasks.python import vision
import os

# --- 設定項目 ---
UV_MAKEUP_DIR = Path("../output_uv_textures/makeup")
# ★パスの確認をお願いします
ORIG_MAKEUP_DIR = Path(r"E:\cosme_subjects\20250110cosme\exp005_20250110_14_31_fundA1_p85") 

ROI_X, ROI_Y, ROI_W, ROI_H = 80, 130, 15, 15
START_F, END_F = 0, 300
OBJ_PATH = "canonical_face_model.obj"
# ----------------

# (get_signal_from_uv_roi, visualize_correlation, load_face_mesh_topology は変更なし)
# ※未定義エラーが出る場合は以前のコードからコピーしてください
# --- 以前のコードの関数が必要であればここに貼り付けてください ---

def sample_colors_from_uv(uv_texture, uv_coords):
    """
    UV展開図(uv_texture)から、468個の頂点座標(uv_coords)にある色を取得する
    """
    h, w = uv_texture.shape[:2]
    vertex_colors = []
    
    for uv in uv_coords:
        u, v = int(uv[0]), int(uv[1])
        # 範囲外ガード
        u = np.clip(u, 0, w - 1)
        v = np.clip(v, 0, h - 1)
        
        # 色を取得 (BGR)
        color = uv_texture[v, u]
        # pythonのリストに入れるため int型キャスト
        vertex_colors.append((int(color[0]), int(color[1]), int(color[2])))
        
    return vertex_colors

def draw_mesh_heat_map(dest_shape, landmarks, triangles, vertex_colors):
    """
    顔の形状に合わせて、三角形ごとに色を塗る
    """
    output_img = np.zeros((dest_shape[0], dest_shape[1], 3), dtype=np.uint8)
    
    for tri in triangles:
        # 三角形の3つの頂点インデックス
        idx1, idx2, idx3 = tri
        
        # 3点の座標
        pt1 = landmarks[idx1]
        pt2 = landmarks[idx2]
        pt3 = landmarks[idx3]
        
        # 3点の色
        c1 = np.array(vertex_colors[idx1])
        c2 = np.array(vertex_colors[idx2])
        c3 = np.array(vertex_colors[idx3])
        
        # 平均色を計算
        mean_color = (c1 + c2 + c3) / 3.0
        draw_color = (int(mean_color[0]), int(mean_color[1]), int(mean_color[2]))
        
        # 三角形を描画 (塗りつぶし)
        # 座標をint型配列に変換
        pts = np.array([pt1, pt2, pt3], np.int32)
        pts = pts.reshape((-1, 1, 2))
        
        # 隙間ができないように少し太めに描画してから中を塗る手もあるが、
        # fillConvexPoly は高速で綺麗
        cv2.fillConvexPoly(output_img, pts, draw_color)
        
    return output_img

# --- メイン実行処理 ---

try:
    # 0. 初期化
    script_dir = os.path.dirname(os.path.abspath("__file__"))
    model_path = "face_landmarker.task"
    if not os.path.exists(model_path): model_path = os.path.join(script_dir, "face_landmarker.task")
    with open(model_path, 'rb') as f: model_buffer = f.read()
    base_options = python.BaseOptions(model_asset_buffer=model_buffer)
    detector = vision.FaceLandmarker.create_from_options(vision.FaceLandmarkerOptions(base_options=base_options, num_faces=1))
    
    # トポロジー読み込み
    uv_coords, triangles = load_face_mesh_topology(OBJ_PATH)

    # 1. 信号抽出 & 内積マップ作成
    uv_signal = get_signal_from_uv_roi(UV_MAKEUP_DIR, START_F, END_F, ROI_X, ROI_Y, ROI_W, ROI_H)
    result_map = visualize_correlation(UV_MAKEUP_DIR, uv_signal, START_F)
    
    # 2. テクスチャ画像作成 (グレースケール)
    gamma = 0.5
    norm_map = (np.clip(result_map, 0, None) / (np.max(np.abs(result_map)) + 1e-6)) ** gamma
    gray_map = (norm_map * 255).astype(np.uint8)
    # ここでは膨張処理などは不要です。元のクリアなデータを使います。
    texture_img = cv2.merge([gray_map, gray_map, gray_map])

    # 3. 元画像読み込み & 顔検出
    orig_files = sorted(list(ORIG_MAKEUP_DIR.glob("*.npy")))
    if not orig_files: raise FileNotFoundError(f"元の画像が見つかりません: {ORIG_MAKEUP_DIR}")
    
    img_orig = np.load(orig_files[0])
    if img_orig.dtype != np.uint8: img_orig = np.clip(img_orig, 0, 255).astype(np.uint8)
    h_orig, w_orig = img_orig.shape[:2]
    
    mp_img = mp.Image(image_format=mp.ImageFormat.SRGB, data=cv2.cvtColor(img_orig, cv2.COLOR_BGR2RGB))
    detection = detector.detect(mp_img)
    
    rewrapped_face = np.zeros_like(img_orig)
    
    if detection.face_landmarks:
        lms = detection.face_landmarks[0]
        landmarks_xy = np.array([(lm.x * w_orig, lm.y * h_orig) for lm in lms])
        
        print("新しいロジックで再マッピング中 (Vertex Sampling)...")
        
        # [Step 1] UVマップ上の468点から色を吸い取る
        vertex_colors = sample_colors_from_uv(texture_img, uv_coords)
        
        # [Step 2] 顔の形に合わせてメッシュを描画 (モザイク状になる)
        mesh_img = draw_mesh_heat_map((h_orig, w_orig), landmarks_xy, triangles, vertex_colors)
        
        # [Step 3] スムージング (ぼかし処理)
        # メッシュのカクカク感を消して、自然なヒートマップにする
        # 顔の大きさによってカーネルサイズ（ぼかし量）は調整してください
        blur_kernel = (15, 15) 
        rewrapped_face = cv2.GaussianBlur(mesh_img, blur_kernel, 0)

    # 4. 合成 (不透明度調整済み)
    blended_face = img_orig.copy()
    
    # マスク作成 (描画された部分のみ)
    mask = (np.sum(rewrapped_face, axis=2) > 0).astype(np.uint8)
    
    # ガウシアンブラーでマスクのエッジも少しぼかすと馴染みが良い
    # mask = cv2.GaussianBlur(mask, (5, 5), 0) # オプション
    
    mask_3ch = cv2.merge([mask, mask, mask])
    
    # 合成比率 (元画像 0.3 : 解析結果 0.7)
    overlay = cv2.addWeighted(img_orig, 0.0, rewrapped_face, 1.0, 0)
    
    # マスク部分のみ書き換え
    # (maskが0か1かのバイナリなので、厳密には境界線でジャギーが出る可能性がありますが、
    #  ヒートマップ用途なら十分許容範囲です)
    np.copyto(blended_face, overlay, where=(mask_3ch > 0))

    # 5. プロット
    plt.figure(figsize=(12, 6))
    plt.subplot(1, 2, 1)
    plt.imshow(texture_img, cmap='gray', vmin=0)
    plt.title("1. UV Map Data")
    plt.axis('off')
    
    plt.subplot(1, 2, 2)
    plt.imshow(cv2.cvtColor(blended_face, cv2.COLOR_BGR2RGB))
    plt.title("2. Reconstructed Heatmap (Mesh Fill)")
    plt.axis('off')
    plt.show()

except Exception as e:
    import traceback
    traceback.print_exc()
    print(f"エラー: {e}")
```

```python
#素肌キャリブレーション
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from tqdm import tqdm

# --- 設定 ---
BASE_DIR = Path("../output_uv_textures")
DIR_BARE = BASE_DIR / "bare_face"
DIR_MAKEUP = BASE_DIR / "makeup"

# 解析するフレーム数 (多いほど振幅の推定が安定します)
START_F = 30
END_F = 330 
# -----------

def get_amplitude_map(img_dir: Path, start_f: int, end_f: int):
    """
    指定ディレクトリの画像群から、各画素ごとの「脈波信号の振幅(標準偏差)」を計算して返します。
    """
    files = sorted(list(img_dir.glob("*.npy")))
    files = files[start_f : min(len(files), end_f)]
    
    if len(files) == 0:
        raise FileNotFoundError(f"画像が見つかりません: {img_dir}")

    # (Time, Height, Width) のデータを蓄積するリスト
    ac_stack = []
    
    # フィルタ用の初期値
    accum_face = None 

    print(f"振幅計算中: {img_dir.name} ({len(files)} frames)")

    for f_path in tqdm(files):
        img = np.load(f_path)
        face_g = img[:, :, 1].astype(np.float64) # Gチャンネル
        
        if accum_face is None:
            accum_face = face_g
            # 初回はAC成分0とする
            ac_component = np.zeros_like(face_g)
        else:
            # ローパスフィルタでDC成分を更新
            accum_face = accum_face * 0.95 + face_g * 0.05
            # 原信号 - DC成分 = AC成分(脈波)
            ac_component = face_g - accum_face
            
        ac_stack.append(ac_component)

    # numpy配列に変換 (Time, H, W)
    ac_stack = np.array(ac_stack)
    
    # 時間方向(axis=0)の標準偏差を計算 -> これを「振幅」とする
    # (H, W) の2次元マップができる
    amplitude_map = np.std(ac_stack, axis=0)
    
    return amplitude_map

# --- 実行プロセス ---

try:
    # 1. 素肌(Bare)の振幅マップを作成
    amp_map_bare = get_amplitude_map(DIR_BARE, START_F, END_F)
    
    # 2. 化粧(Makeup)の振幅マップを作成
    amp_map_makeup = get_amplitude_map(DIR_MAKEUP, START_F, END_F)
    
    # 3. 振幅比の計算 (Makeup / Bare)
    # ゼロ除算を防ぐため、Bareの振幅が極端に小さい場所(背景など)はマスクする
    min_amp_threshold = 0.1 # ノイズフロア以下の振幅は無視
    
    valid_mask = amp_map_bare > min_amp_threshold
    
    ratio_map = np.zeros_like(amp_map_bare)
    
    # 有効な領域だけ割り算を実行
    # Ratio = 化粧時の振幅 / 素肌時の振幅
    ratio_map[valid_mask] = amp_map_makeup[valid_mask] / amp_map_bare[valid_mask]

    # --- 可視化 ---
    plt.figure(figsize=(15, 5))

    # 素肌の振幅
    plt.subplot(1, 3, 1)
    plt.imshow(amp_map_bare, cmap='inferno', vmin=0)
    plt.title("Amplitude: Bare Skin")
    plt.axis('off')
    plt.colorbar(fraction=0.046, pad=0.04)

    # 化粧の振幅
    plt.subplot(1, 3, 2)
    plt.imshow(amp_map_makeup, cmap='inferno', vmin=0)
    plt.title("Amplitude: Makeup")
    plt.axis('off')
    plt.colorbar(fraction=0.046, pad=0.04)

    # 振幅比 (Ratio Map)
    plt.subplot(1, 3, 3)
    # 1.0付近が「変化なし」、1.0以下が「減衰」
    # 見やすくするため 0.0 ~ 1.5 程度でクリップして表示
    plt.imshow(ratio_map, cmap='jet', vmin=0.0, vmax=1.5)
    plt.title("Ratio (Makeup / Bare)\nBlue<1.0 (Attenuated), Red>1.0 (Noise?)")
    plt.axis('off')
    plt.colorbar(fraction=0.046, pad=0.04, label="Amplitude Ratio")

    plt.tight_layout()
    plt.show()

    # (オプション) 保存
    # output_path = BASE_DIR / "ratio_map.npy"
    # np.save(str(output_path), ratio_map)
    # print(f"Ratio map saved to {output_path}")

except Exception as e:
    print(f"エラーが発生しました: {e}")
```
